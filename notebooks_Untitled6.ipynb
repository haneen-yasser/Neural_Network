```python
import numpy as np

# tanh activation function
def tanh(x):
    return np.tanh(x)

# Input
x = np.array([[0.6],
              [0.1]])

# Random weights in [-0.5, 0.5]
W1 = np.random.uniform(-0.5, 0.5, (2, 2))
W2 = np.random.uniform(-0.5, 0.5, (1, 2))

# Bias values
b1 = 0.5
b2 = 0.7

# Forward propagation
z1 = np.dot(W1, x) + b1
a1 = tanh(z1)

z2 = np.dot(W2, a1) + b2
output = tanh(z2)

print("Network Output:")
print(output)
```
